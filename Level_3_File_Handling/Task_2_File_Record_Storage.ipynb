# Task 2: Save and Retrieve Records Using a File


```python
# Store student records
records = []

n = int(input("Enter number of students: "))

for i in range(n):
    print("\nEnter details for student", i + 1)

    name = input("Name: ")
    age = int(input("Age: "))
    course = input("Course: ")

    record = {
        "Name": name,
        "Age": age,
        "Course": course
    }

    records.append(record)

# Save records to a file
with open("student_records.txt", "w") as file:
    for record in records:
        file.write(str(record) + "\n")

print("\nRecords saved successfully.")

# Retrieve records from the file
print("\nSaved Student Records:")

with open("student_records.txt", "r") as file:
    for line in file:
        print(line.strip())
```


## Objective
Save records into a file and retrieve them when required.

## Description
The program stores student records in a file and then reads
the saved records from the file.
