# Task 2: Check Whether a Given Number Exists in the Grid

```python
grid = [ [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]]

number = int(input("Enter a number: "))

found = False

for row in grid:
    for value in row:
        if value == number:
            found = True

if found:
    print(number, "exists in the grid.")
else:
    print(number, "does not exist in the grid.")
print(grid)    
```


## Objective
Check whether a given number exists in a predefined 3×3 number grid.

## Description
The program takes a number from the user and searches for that number
in the predefined grid. It displays whether the number exists or not.

## Concepts Used
- Lists
- Nested for loops
- User input
- Conditional statements
- Boolean variable
