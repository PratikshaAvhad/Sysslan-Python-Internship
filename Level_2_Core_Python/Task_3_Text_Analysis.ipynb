# Task 3: Text Character and Word Count


```python

text = input("Enter a text: ")

# Count characters
character_count = len(text)

# Count words
word_count = len(text.split())

print("Number of characters:", character_count)
print("Number of words:", word_count)
```


## Objective
Count the number of characters and words in a given text.

## Description
The program takes text as input from the user and calculates
the total number of characters and words in the text.

## Concepts Used
- Strings
- User input
- `len()` function
- `split()` method
- String manipulation
