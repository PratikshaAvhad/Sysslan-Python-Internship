# Task 1: Record Management

```python
# Store multiple records
records = []

# Get number of records
n = int(input("Enter number of students: "))

# Add records
for i in range(n):
    print("\nEnter details for student", i + 1)

    name = input("Name: ")
    age = int(input("Age: "))
    course = input("Course: ")

    record = {
        "Name": name,
        "Age": age,
        "Course": course
    }

    records.append(record)

# Display all records
print("\nStudent Records:")

for record in records:
    print(record)
```


## Objective
Create and manage multiple records using Python.

## Description
The program allows the user to add multiple student records
and display the stored records.

## Concepts Used
- Dictionaries
- Lists
- User input
- `append()` method
- Loops
