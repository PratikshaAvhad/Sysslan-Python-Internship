# Task 1: Email Validation


```python
import re

email = input("Enter your email address: ")

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

if re.match(pattern, email):
    print("Valid email address.")
else:
    print("Invalid email address.")
```


## Objective
Validate whether a given email address follows a valid email format.

## Description
The program takes an email address from the user and validates it
using a Regular Expression (Regex) pattern. The pattern checks the
basic structure of an email address, including the username, @ symbol,
domain name, and domain extension.

## Concepts Used
- Regular Expressions (Regex)
- `re` module
- String validation
- User input
- Conditional statements
- Pattern matching
