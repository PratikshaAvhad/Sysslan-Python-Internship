# Task 4: Numerical Data Analysis

```python
# Create an empty list
numbers = []

# Get number of values
n = int(input("How many numbers do you want to enter? "))

# Add numbers to the list
for i in range(n):
    number = float(input(f"Enter number {i + 1}: "))
    numbers.append(number)

# Calculate numerical values
total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)

# Display results
print("\nNumerical Data Analysis")
print("Numbers:", numbers)
print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
```


## Objective
Analyze a list of numbers and calculate basic numerical values.

## Description
The program takes numbers from the user and calculates the total,
average, and maximum value from the given numbers.

## Concepts Used
- Lists
- User input
- Loops
- Arithmetic operations
- Built-in functions

## Important Functions Used
- `input()` → takes numbers from the user
- `append()` → adds numbers to the list
- `sum()` → calculates the total
- `len()` → counts the numbers
- `max()` → finds the maximum value
