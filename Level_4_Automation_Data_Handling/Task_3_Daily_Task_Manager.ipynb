# Task 3: Daily Task List Manager

```python
# Create an empty task list
tasks = []

# Get number of tasks
n = int(input("How many tasks do you want to add? "))

# Add tasks
for i in range(n):
    task = input(f"Enter task {i + 1}: ")
    tasks.append(task)

# Display tasks
print("\nDaily Task List:")

for i, task in enumerate(tasks, start=1):
    print(f"{i}. {task}")

# Save tasks to a file
with open("daily_tasks.txt", "w") as file:
    for task in tasks:
        file.write(task + "\n")

print("\nTasks saved successfully.")
```


## Objective
Create a simple daily task list and save the tasks to a file.

## Description
The program takes daily tasks from the user, stores them in a list,
and saves all tasks into a text file.

## Concepts Used
- Lists
- User input
- Loops
- File handling
- Functions
- String handling

## Important Functions Used
- `input()` → takes task from the user
- `append()` → adds a task to the list
- `open()` → creates or opens a file
- `write()` → saves tasks into the file

## Before Running
No external file is required.

The program will create `daily_tasks.txt`
when the program is executed.
