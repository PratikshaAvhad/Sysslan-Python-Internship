# Task 3: Automated Backup


```python
import os
import shutil

source_folder = "data"
backup_folder = "backup"

# Create backup folder
os.makedirs(backup_folder, exist_ok=True)

# Copy files to backup folder
for file in os.listdir(source_folder):
    source_path = os.path.join(source_folder, file)
    backup_path = os.path.join(backup_folder, file)

    if os.path.isfile(source_path):
        shutil.copy(source_path, backup_path)

print("Backup completed successfully.")
```


## Objective
Create a backup of files automatically using Python.

## Description
The program copies files from a source folder to a backup folder.
This helps protect important files by keeping a separate copy.

## Concepts Used
- File handling
- `os` module
- `shutil` module
- Folder creation
- File copying
- Loops
- Automation

## Important Functions Used
- `os.listdir()` → folder files
- `os.makedirs()` → backup folder 
- `os.path.join()` → file path
- `os.path.isfile()` → given path file check
- `shutil.copy()` → file copy
