# Task 2: CSV Data Reader

```python
import csv

# Open the CSV file
with open("students.csv", "r") as file:
    reader = csv.DictReader(file)

    print("Student Records\n")

    # Display student information
    for row in reader:
        print("Name:", row["student_name"])
        print("Department:", row["department"])
        print("City:", row["city"])
        print("Marks:", row["marks"])
        print("-" * 30)
```



## Objective
Read student data from a CSV file and display it using Python.

## Description
The program reads student records from a CSV file and displays
important information such as name, department, city, and marks.

## Concepts Used
- CSV files
- File handling
- CSV module
- Dictionaries
- Loops

## Important Functions Used
- `open()` → opens the CSV file
- `csv.DictReader()` → reads CSV data as dictionaries
- `print()` → displays the data
