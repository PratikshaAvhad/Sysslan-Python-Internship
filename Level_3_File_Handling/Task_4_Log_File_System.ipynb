# Task 4: Create a Log File with Date and Time

```python
from datetime import datetime

# Get current date and time
current_time = datetime.now()

message = input("Enter log message: ")

# Add message to the log file
with open("activity_log.txt", "a") as file:          # a means append mode
    file.write(f"{current_time} - {message}\n")

print("Log saved successfully.")
```



## Objective
Create a log file that stores activities along with the current date and time.

## Description
The program records a message in a log file with the current
date and time.

## Concepts Used
- File handling
- `datetime` module
- Date and time
- Append mode
- `with open()`
