# Task 4: Personal Utility Tool

```python
from datetime import datetime

def show_datetime():
    current_time = datetime.now()
    print("Current Date and Time:", current_time)

def calculate_square():
    number = float(input("Enter a number: "))
    print("Square:", number ** 2)

def celsius_to_fahrenheit():
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9 / 5) + 32
    print("Temperature in Fahrenheit:", fahrenheit)

while True:
    print("\nPersonal Utility Tool")
    print("1. Current Date and Time")
    print("2. Calculate Square")
    print("3. Celsius to Fahrenheit")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_datetime()

    elif choice == "2":
        calculate_square()

    elif choice == "3":
        celsius_to_fahrenheit()

    elif choice == "4":
        print("Thank you for using the utility tool.")
        break

    else:
        print("Invalid choice. Please try again.")
```



## Objective
Create a simple personal utility tool using Python.

## Description
The program provides useful utilities such as checking the
current date and time, calculating the square of a number,
and converting temperature from Celsius to Fahrenheit.

## Concepts Used
- Functions
- User input
- Conditional statements
- Date and time
- Arithmetic operations
- Menu-driven program

## Important Functions Used
- `datetime.now()` → gets the current date and time
- `input()` → takes input from the user
- `float()` → converts input into a decimal number
- `print()` → displays the result

## Before Running
No external library or file is required.
