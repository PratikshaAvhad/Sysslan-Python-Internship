# Task 2: Random Password Generator

```python
import random
import string

length = int(input("Enter password length: "))

characters = string.ascii_letters + string.digits + string.punctuation

password = ""

for i in range(length):
    password += random.choice(characters)

print("Generated Password:", password)
```



## Objective
Generate a random and secure password using Python.

## Description
The program generates a random password by combining letters,
numbers, and special characters. The user can specify the desired
password length.

## Concepts Used
- random module
- String manipulation
- User input
- Loops
- `random.choice()`
- Character sets
