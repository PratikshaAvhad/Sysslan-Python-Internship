# Task 2: Student Report Generator

```python
import tkinter as tk
from tkinter import messagebox

# Generate student report
def generate_report():
    try:
        name = name_entry.get()

        marks = [
            float(subject1_entry.get()),
            float(subject2_entry.get()),
            float(subject3_entry.get())
        ]

        # Calculate total and average
        total = sum(marks)
        average = total / len(marks)

        # Check result
        if average >= 40:
            result = "Pass"
        else:
            result = "Fail"

        # Store report data
        report = {
            "Name": name,
            "Marks": marks,
            "Total": total,
            "Average": average,
            "Result": result
        }

        # Display report
        report_label.config(
            text=f"Student Report\n\n"
                 f"Name: {report['Name']}\n"
                 f"Marks: {report['Marks']}\n"
                 f"Total: {report['Total']}\n"
                 f"Average: {report['Average']:.2f}\n"
                 f"Result: {report['Result']}"
        )

    except ValueError:
        messagebox.showerror("Invalid Input", "Please enter valid marks.")


# Create main window
window = tk.Tk()
window.title("Student Report Generator")
window.geometry("400x500")

# Title
title_label = tk.Label(
    window,
    text="Student Report Generator",
    font=("Arial", 18, "bold")
)
title_label.pack(pady=20)

# Student name
tk.Label(window, text="Student Name").pack()
name_entry = tk.Entry(window, width=30)
name_entry.pack(pady=5)

# Subject 1
tk.Label(window, text="Subject 1 Marks").pack()
subject1_entry = tk.Entry(window, width=30)
subject1_entry.pack(pady=5)

# Subject 2
tk.Label(window, text="Subject 2 Marks").pack()
subject2_entry = tk.Entry(window, width=30)
subject2_entry.pack(pady=5)

# Subject 3
tk.Label(window, text="Subject 3 Marks").pack()
subject3_entry = tk.Entry(window, width=30)
subject3_entry.pack(pady=5)

# Generate button
generate_button = tk.Button(
    window,
    text="Generate Report",
    command=generate_report
)
generate_button.pack(pady=20)

# Report output
report_label = tk.Label(
    window,
    text="Enter details and click Generate Report",
    font=("Arial", 12),
    justify="left"
)
report_label.pack(pady=10)

# Run GUI
window.mainloop()
```
