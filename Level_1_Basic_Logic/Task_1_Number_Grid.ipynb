# Task 1: Display a Simple 3×3 Number Grid

```python
for i in range(1, 10):
    print(i, end=" ")
    if i % 3 == 0:
        print()
```


## Objective
Display a simple 3×3 number grid using Python.

## Description
The program uses a loop to display numbers from 1 to 9
in a structured 3×3 grid format.

## Concepts Used
- for loop
- range()
- Conditional statements
- Modulus operator (%)
- print() function
- Structured output
