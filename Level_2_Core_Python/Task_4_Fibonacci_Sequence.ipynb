# Task 4: Fibonacci Sequence


```python

terms = int(input("Enter the number of terms: "))

# First two Fibonacci numbers
a = 0
b = 1

print("Fibonacci Sequence:")

# Generate the sequence
for i in range(terms):
    print(a, end=" ")
    a, b = b, a + b
```


## Objective
Generate and display a Fibonacci sequence for a given number of terms.

## Description
The program takes the number of terms from the user and generates
the Fibonacci sequence using a loop.

- Variables
- User input
- for loop
