# Task 4: Verify All Numbers Are Unique

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

numbers = []

unique = True

for row in grid:
    for value in row:
        if value in numbers:
            unique = False
        else:
            numbers.append(value)

if unique:
    print("All numbers in the grid are unique.")
else:
    print("The grid contains duplicate numbers.")
```


## Objective
Verify whether all numbers in a predefined 3×3 number grid are unique.

## Description
The program checks the numbers in the predefined grid and verifies
whether any number is repeated. It displays whether all numbers
are unique or not.

## Concepts Used
- Lists
- Nested for loops
- Conditional statements
- Boolean variable
- Membership operator (in)
- Duplicate value checking
