# Task 1: Web Scraping using BeautifulSoup


```python
import requests
from bs4 import BeautifulSoup

# Get webpage content
url = "https://example.com"
response = requests.get(url)

# Parse HTML content
soup = BeautifulSoup(response.text, "html.parser")

# Extract heading
heading = soup.find("h1")

print("Webpage Heading:")

if heading:
    print(heading.get_text())
else:
    print("No heading found.")
```


## Objective
Extract useful information from a webpage using Python.

## Description
The program uses BeautifulSoup to read HTML content
and extract text from the webpage.

## Concepts Used
- Web scraping
- BeautifulSoup
- Requests
- HTML
- Lists
- Loops

## Important Functions Used
- `requests.get()` → gets webpage content
- `BeautifulSoup()` → reads and parses HTML
- `find_all()` → finds matching HTML elements
- `.get_text()` → extracts text from an HTML element
