# Task 3: Read File Content Line by Line

```python
# Create a sample text file
with open("sample.txt", "w") as file:
    file.write("Python Programming\n")
    file.write("File Handling\n")
    file.write("Data Processing\n")

# Read the file line by line
with open("sample.txt", "r") as file:
    for line in file:
        print(line.strip())
```



## Objective
Read and display the contents of a file line by line.

## Description
The program reads a text file safely and displays each line
separately.

## Concepts Used
- File handling
- Read mode
- `with open()`
- for loop
- Line-by-line reading
